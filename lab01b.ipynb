```python
order_total = float(input("Enter the order total: "))
day_of_week = str(input("Enter the day of the week: ")).capitalize()
if order_total < 49.99 and day_of_week not in ["Saturday", "Sunday"]:
    Delivery_charge = 8
    print(f"Your delivery charge is: ${Delivery_charge:}")
elif order_total < 49.99 and day_of_week in ["Saturday", "Sunday"]:
    Delivery_charge = 11
    print(f"Your delivery charge is: ${Delivery_charge:}")
elif order_total >= 49.99 and day_of_week in ["Monday","Tuesday","Wednesday","Thursday","Friday"]:
    Delivery_charge = 4
    print(f"Your delivery charge is: ${Delivery_charge:}")
elif order_total >= 49.99 and day_of_week in ["Saturday","Sunday"]:
    Delivery_charge = 6
    print(f"Your delivery charge is: ${Delivery_charge:}")
```
